Cell 1: library


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

Cell 2: Data Sintesis

```python
np.random.seed(2008)
n_samples = 700

X1 = np.random.rand(n_samples) * 3
X2 = np.random.rand(n_samples) * 3

noise = np.random.rand(n_samples) * 1.5

#Pengaruh linear: y = 5 + 2.5*X1 - 2 *X2 + noise
y = 5 + 2.5 * X1 - 2 * X2 + noise

#Menggabungkan menjadi data frame

data = pd.DataFrame({'X1': X1, 'X2': X2, 'y': y})
display(data.head())
```

Cell 3: Model fit

```python
# Memisahkan fitur (X) dan target (y)
X = data[['X1', 'X2']]
y = data['y']

# Membagi data menjadi train set dan test set, training 80% testing 20%
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2008)

# Melakukan fitting model
model = LinearRegression()
model.fit(X_train, y_train)



print(f"Intercept (koefisien bias): {model.intercept_:.4f}")
print(f"Koefisien X1: {model.coef_[0]:.4f}")
print(f"Koefisien X2: {model.coef_[1]:.4f}")


# y = 5.79 + 2.49 X1 - 2.02 X2
```

Cell 4: Prediksi dan Matriks

```python
#Prediksi
y_pred = model.predict(X_test)

#Matriks
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

#Print matriks
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"R-squared (R2 Score): {r2:.4f}")
```

Cell 5: Visualisasi

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(x=y_test, y = y_pred, label='Data Asli', color='blue', alpha = 0.7)
plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], linestyle='--', color='red', label='Prediksi')
plt.title('Nilai aktual vs prediksi')
plt.xlabel('Nilai aktual (Y)')
plt.ylabel('Nilai prediksi (Y_topi)')
plt.legend()
plt.grid(True)
plt.show()
```
